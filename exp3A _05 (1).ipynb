```python
import numpy as np
from queue import PriorityQueue

def a_star_search(graph,start,goal,h):
    open_set=PriorityQueue()
    open_set.put((0,start)) # (f_score, node)

    g_score={node:float('inf') for node in graph}
    g_score[start]=0

    came_from={} # Parent node in the path

    while not open_set.empty():
        current_f, current = open_set.get()

        # If we have reached the goal, reconstruct and return the path
        if current == goal:
            path = []
            while current in came_from:
                path.append(current)
                current = came_from[current]
            path.append(start) # Add the start node
            path.reverse()
            return path, g_score[goal]

        # Explore neighbors
        for neighbor, edge_cost in graph[current]:
            # Calculate the cost to reach the neighbor through the current node
            tentative_g_score = g_score[current] + edge_cost

            if tentative_g_score < g_score[neighbor]:
                g_score[neighbor] = tentative_g_score
                f_score = tentative_g_score + h(neighbor)
                open_set.put((f_score, neighbor))
                came_from[neighbor] = current

    return None, float('inf') # Goal not reachable

graph={
         'A':[('B',1),('C',4)],
         'B':[('A',1),('C',2),('D',5)],
         'C':[('A',4),('B',2),('D',1)],
         'D':[('B',5),('C',1)]
}

heuristic={
    'A':7,
    'B':6,
    'C':2,
    'D':0
}

def h(node):
    return heuristic[node]

path, cost= a_star_search(graph,'A','D',h)

print("path:",path)
print("cost:",cost)
```
